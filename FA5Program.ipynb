```python
#FA5 Program

def calculate_total(price_per_item, quantity):
    total = price_per_item * quantity
    return total

try:
    item_price = float(input("Enter the price: "))
    items_quantity = int(input("Enter the quantity: "))

    final_total_price = calculate_total(item_price, items_quantity)

    print(f"The total price is of your purchases is: ₱{final_total_price:.2f}")
except ValueError:
    print("Invalid input, Please try again.")
```
